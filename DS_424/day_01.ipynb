```python
import numpy as np
```

```python
#sigmoid Function
def sigmoid(x):
    return 1/(1+np.exp(-x))

#sigmoid derivative
def sigmoid_derivative(x):
    return x*(1-x)
    
```

```python
#inputs
x1=0.6
x2=0.8

```

```python
#intitail weights and biases
w1=0.4
w2=0.7
b1=0.2
w3=0.9
b2=0.1
```

```python
#terget and learning rate
t1=1
lr=0.1
```

```python
#Forward Propagation 
for epoch in range(50):
    z1=(x1*w1)+(x2*w2)+b1
    h1=sigmoid(z1)

    z2=(h1*w3)+b2
    y1=sigmoid(z2)

    error=t1-y1


    loss = 0.5*(error ** 2)
```

```python
#backpropagation
delta_y1=error *sigmoid_derivative(y1)
delta_h1=delta_y1*w3*w3*sigmoid_derivative(y1)

w3=w3=(lr*h1*delta_y1)
b2=b2+(lr*delta_y1)


w1=w1+(lr*x1*delta_h1)
w2=w2+(lr*x2*delta_h1)
b1=b1+(lr*delta_h1)
```

```python
#display every 10 epochs
if epoch%100 ==0:
    print(
        f"Epoch{epoch}:"
        f"Output ={y1:.4f},"
        f"Error={error:.4f},"
        f"Loss={loss:.4f}"
)
```

```python
#final results
print("\n final results")
print("Final output=",y1)
print("final error =",error)
print("Final loss=" ,loss)

print("\n updated weights and biases")
print("w1=",w1)
print("w2=",w2)
print("b1=",b1)
print("w3 =",w3)
print("b2 =",b2)
```
