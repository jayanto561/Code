```python
import heapq 
```

```python
class Node:
    """A node in the search grid, used for path reconstruction."""
    def __init__(self, position, g=0, h=0, parent=None):
        self.position = position 
        self.g = g                
        self.h = h                 
        self.f = g + h             
        self.parent = parent
```

```python
    def __lt__(self, other):
        return self.f < other.f
```

```python
def manhattan_distance(a, b):
    """Admissible heuristic for 4-directional grid movement."""
    return abs(a[0] - b[0]) + abs(a[1] - b[1])
```

```python
def get_neighbors(position, grid):
    """Return valid, unblocked, in-bounds neighbors (4-directional)."""
    rows, cols = len(grid), len(grid[0])
    r, c = position
    candidates = [(r - 1, c), (r + 1, c), (r, c - 1), (r, c + 1)]
    neighbors = []
    for nr, nc in candidates:
        if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 0:
            neighbors.append((nr, nc))
    return neighbors
```

```python
def print_grid_with_path(grid, path, start, goal):
    """Text visualization: # obstacle, S start, G goal, * path, . free cell."""
    path_set = set(path) if path else set()
    for r, row in enumerate(grid):
        line = ""
        for c, cell in enumerate(row):
            pos = (r, c)
            if pos == start:
                line += "S "
            elif pos == goal:
                line += "G "
            elif cell == 1:
                line += "# "
            elif pos in path_set:
                line += "* "
            else:
                line += ". "
        print(line)
```
