```python
import re
custom_stopwords = {"no","not","nor","never","don","don't","doesn","doesn't",
                        "didn","didn't","isn","isn't","aren","aren't","wasn",
                      "wasn't","weren","weren't","won","won't","wouldn","wouldn't",
                      "couldn","couldn't","shouldn","shouldn't","shan","shan't",
                      "ain","ain't","haven","haven't","hasn","hasn't","hadn","hadn't",
                      "can","can't","cannot"}
corpus=[]
ps=PorterStemmer()
stop_words=set(stopwords.words("english"))-custom_stopwords
for i in range(len(df)):
    review = re.sub("[^a-zA-Z]", " ", df["Review"][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word)for word in review if word not in stop_words]
    review=" ".join(review)
    corpus.append(review)
```

```python
df["processed_text"]=corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=1500)
```

```python
x=cv.fit_transform(corpus).toarray()
```

```python
x.shape
```

```python
y =df["Liked"]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
nb  =GaussianNB()
nb.fit(x_train,y_train)
y_pred=nb.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.linear_model import LogisticRegression
lr  = LogisticRegression()
lr.fit(x_train,y_train)
y_pred=lr.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf  = RandomForestClassifier()
rf.fit(x_train,y_train)
y_pred=rf.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
import joblib
```

```python
joblib.dump(lr,"Restaurant_review_model")
```

```python
def predict_feedback(raw_text):
    # 1. Remove non-letters (same as your training cleaning)
    review = re.sub("[^a-zA-Z]", " ", raw_text)

    # 2. Lowercase

    review = review.lower()

    # 3. Split into words
    review = review.split()

    # 4. Remove stopwords (but keep negation words, same as training)
    review = [ps.stem(word) for word in review if word not in stop_words]

    # 5. Join back into a single string
    review = " ".join(review)

    # 6. Convert to the SAME CountVectorizer features (don't refit — use transform only)
    vec = cv.transform([review]).toarray()

    # 7. Predict using your trained Logistic Regression model
    prediction = lr.predict(vec)[0]

    return "Positive" if prediction == 1 else "Negative"
```

```python
# =========================================================
# Interactive input — type your own review to test
# =========================================================
user_review = input("Enter a customer review: ")
print("Predicted sentiment:", predict_feedback(user_review))
```
