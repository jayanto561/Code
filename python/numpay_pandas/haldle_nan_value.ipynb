```python
from sklearn.impute import SimpleImputer
```

```python
df2=df.copy()
df2
```

```python
#fixed change
impute=SimpleImputer(strategy="constant",fill_value=-1)
df2["a"]=impute.fit_transform(df2[["a"]])
df2
```

```python
impute=SimpleImputer(strategy="mean")
df2["d"]=impute.fit_transform(df2[["d"]])
df2
```

```python
impute=SimpleImputer(strategy="median")
df2["d"]=impute.fit_transform(df2[["d"]])
df2
```

```python
dff={"A":[1,2,np.nan,3,np.nan],
     "B":[np.nan,np.nan,np.nan,np.nan,np.nan],
    "C":[11,12,13,14,15],
    "D":[16,18,np.nan,np.nan,18]
    
}
```

```python
df3=pd.DataFrame(dff)
df3
```

```python
impute=SimpleImputer(strategy="most_frequent")
df3["D"]=impute.fit_transform(df3[["D"]])
df3
```

```python
from sklearn.impute import KNNImputer
```

```python
df3=df3.copy()
df3
```

```python
knn=KNNImputer(n_neighbors=3,weights="uniform")
df3=knn.fit_transform(df3)
df3
```

```python
d={
    "store":["walton","walton","volvo","volvo","singer","singer"],
    "customer":["jayanto","sabit","kabbo","tanin","ahim","fuad"],
    "sales":[150,120,300,220,430,110]
}
df=pd.DataFrame(d)
df
```

```python
#group by store sum of sales
df.groupby("store")[["sales"]].sum()
```

```python
df.groupby("store")[["sales"]].mean()
```

```python
df.groupby("store")[["sales"]].median()
```
