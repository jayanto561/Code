```python
import pandas as pd
import numpy as np
```

```python
df_1=pd.DataFrame({"key":["a","b","c","d","e"],
                    "A":range(5),
                    "B":range(5,10)
    })
```

```python
df_1
```

```python
df_2=pd.DataFrame({"key":["b","d","f"],
                   "C":range(3),
                   "D":range(3,6)
                  })
```

```python
df_2
```

```python
pd.merge(df_1,df_2,how='inner',on='key')
```

```python
pd.merge(df_1,df_2,how='inner',on='key')[['key','A','D']]
```

```python
pd.merge(df_1,df_2,how="left",on="key")
```

```python
pd.merge(df_1,df_2,how="right",on="key")
```

```python
pd.merge(df_1,df_2,how="outer",on="key")
```
